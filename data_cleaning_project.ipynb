
## Step 1: Load and Inspect Dataset

```python
import pandas as pd

df = pd.read_csv('Messy_Employment_India_Dataset.csv')
print(df.head())
```


## Step 2: Understand Dataset Structure

```python
print("Rows and Columns:", df.shape)
print(df.columns)
df.info()
df.describe(include="all")
```


## Step 3: Handle Missing Values

```python


df = pd.read_csv("Messy_Employment_India_Dataset.csv")
print(df.isnull().sum())
#print(df.isnull().sum().sum())
#print(df[df.isnull().any(axis=1)])
```


## Step 4: Remove Duplicate Records

```python
print("Duplicates:", df.duplicated().sum())

df = df.drop_duplicates()

print("After Removal:", df.duplicated().sum())
```

##Step 5:Correct inconsistent formatting and text values

```python
df["Status"] = df["Status"].str.strip().str.title()
df["Education"] = df["Education"].str.strip().str.title()
df["Industry"] = df["Industry"].str.strip().str.title()
df["Location"] = df["Location"].str.strip().str.title()
df["AI Risk"] = df["AI Risk"].str.strip().str.title()

# Standardize Age Group values
df["Age Group"] = df["Age Group"].str.replace("_", "-", regex=False)

# Check cleaned values
print(df.head())
```


## Step 6: Fix Data Types

```python
df['Date Recorded'] = pd.to_datetime(
    df['Date Recorded']
)

df.info()
```

# Step 7:Detect and handle outliers

```python
from scipy import stats
df = df[(abs(stats.zscore(df.select_dtypes(include='number'))) < 3).all(axis=1)]

print(df.shape)
```


## Step 8: Standardize Column Names

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.lower().str.replace(" ", "_")
print(df.columns)
```

## Step 9: Validate Cleaned Data

```python
print(df.info())

#  missing values
print(df.isnull().sum())

#  duplicate records
print(df.duplicated().sum())

#  first 5 rows
(df.head())
```


## Step 10: Export Cleaned Dataset

```python
df.to_csv("cleaned_dataset.csv", index=False)
print("Cleaned dataset exported successfully!")
```
