## Merging and joining Dataframes
________________________________________________________________________________

## Why do we merge or join data
- Combine related datasets
- perform Database-style Operations
- Analyze data from multiple sources
- Clean and organize the large data
- (This is common in real-world scenarious like sales data ,employee records, and survey results.)
__________________________________________________________________________

## Ways to combine the datasets
pandas provides 3 methords
- merge() column based
- join() index based
- concat() stack based
_______________________________________________________________________________

## merge() Function
it is used in the sqls style
- syntax pd.merge(df1,df2,on='column=1,column=2')
- it allows
- joining on columns
/flexible joins/ handling of overlapping
______________________________________________________________________

```python
import pandas as pd
df1= pd.DataFrame({'A':[1,2,3,4],'B':['a','b','c','d']})
df2= pd.DataFrame({'A':[1,2,3,4],'D':['e','f','g','h']})               #MERGING    (left is inner right is outer)

pd.merge(df1,df2)
```

```python

df1= pd.DataFrame({'A':[1,2,3,4],'B':['a','b','c','d']})
df2= pd.DataFrame({'A':[1,2,3,4],'C':['e','f','g','h']})
pd.merge(df1,df2,on='A' , how="inner",indicator=True )                 # we use indicator to find either merge is use or not
```

```python
df1= pd.DataFrame({'A':[1,2,3,4],'B':['a','b','c','d']})
df2= pd.DataFrame({'A':[1,2,3,4],'C':['e','f','g','h']})
pd.merge(df1,df2,on='A' , how="outer",indicator=True  )
```

```python
df1= pd.DataFrame({'A':[1,2,3,4],'B':['a','b','c','d']})
df2= pd.DataFrame({'A':[1,5,7,3],'C':['e','f','g','h']})
pd.merge(df1,df2,on='A' , how="left",indicator=True   )
```

```python
df1= pd.DataFrame({'A':[1,2,3,4],'B':['a','b','c','d']})
df2= pd.DataFrame({'A':[1,2,3,4],'C':['e','f','g','h']})
pd.merge(df1,df2,on='A' , how="right" )
```

```python
df1= pd.DataFrame({'A':[1,2,3,4],'B':['a','b','c','d']})
df2= pd.DataFrame({'A':[1,2,3,4],'B':['e','f','g','h']})
pd.merge(df1,df2 ,left_index=True, right_index=True)                # we not use on in this code
```

```python
df1= pd.DataFrame({'A':[1,2,3,4],'B':['a','b','c','d']})
df2= pd.DataFrame({'A':[1,2,3,4],'C':['e','f','g','h']})
pd.merge(df1,df2 ,left_index=True,suffixes=("2019","2020"), right_index=True)
```

## Joining
join() is used to combine two Dataframes based on their index
- it is simple yhan merge() when data is already indexed
   - mainly used when:
      - one dataframe is primary
      - another dataframe adds extras column
- syntax
 - df1.join(df2,how='left')
    - df1=left dataframe
    - df2=right dataframe
    - how = type     
    __________________________________________________________________________

## Multiple DataFrames join
    df1.join([df2,df3])

```python
import pandas as pd
d1= pd.DataFrame({'A':[1,2,3,4],'B':['a','b','c','d']})
d2= pd.DataFrame({'C':[1,2,3,4],'D':['e','f','g','h']})
d1.join(d2)
```

```python
d1= pd.DataFrame({'A':[1,2,3,4],'B':['a','b','c','d']})
d2= pd.DataFrame({'C':[1,2,3,4],'D':['e','f','g','h']})
d2.join(d1)
```

## Concatination

```python
import pandas as pd
d1= pd.DataFrame({'A':[1,2,3,4],'B':['a','b','c','d']})
d2= pd.DataFrame({'A':[1,2,3,4],'B':['e','f','g','h']})
pd.concat([d1,d2],axis=0)
```

```python
d1= pd.DataFrame({'A':[1,2,3,4],'B':['a','b','c','d']})
d2= pd.DataFrame({'C':[1,2,3,4],'D':['e','f','g','h']})
pd.concat([d1,d2],axis=0)
```
