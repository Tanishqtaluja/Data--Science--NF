```python
import pandas as pd


```

```python
l1 = [2,4,6,8,10]
var = pd.Series(l1)
print(var)
print(type(var))
```

```python
l1 = [2,4,6,8,10]
var = pd.Series(l1,index = ['a','b','c','r','y'])                     # in this we will give the indexing name as a b c r y
print(var)
```

```python
l1 = [2,4,6,8,10]
var = pd.Series(l1,index = ['a','b','c','r','y'],dtype ="float")           #in this we cahnge the data type
print(var)
```

```python
l1 = [2,4,6,8,10]
var = pd.Series(l1,index = ['a','b','c','r','y'],dtype ="float" ,name = "mylist")       # in this we will change the list name
print(var)
```

```python
var2 = pd.Series(5,index=[1,2,3,4,5])
print(var2)
```

```python
var2 = pd.Series(5,index=[2,4,5,7,8])                                # add both the var having the same index no
var3 = pd.Series(10,index=[2,3,4,5,6])
print(var2+var3)
```
